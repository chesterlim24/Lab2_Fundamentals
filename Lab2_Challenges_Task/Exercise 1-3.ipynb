```python
# [EXERCISE 1]
target_name = "LIM"
SEED_NUM = 4
redacted_name = ""
log_tracker = []

for char in target_name:
    if char in "AEIOU":
        redacted_name += str(SEED_NUM)
        log_tracker.append(f"{char}->{SEED_NUM}")
    else:
        redacted_name += char
        log_tracker.append(f"{char}->{char}")

exec_log = ", ".join(log_tracker)

print(f"Original Surname: {target_name}")
print(f"SEED_NUM Used: {SEED_NUM}")
print(f"Redacted Output: {redacted_name}")
print(f"Execution Log: {exec_log}")
```

```python
# [EXERCISE 2]
engineer_name = "Lim"
SEED_NUM = 4
loops = 7

first_letter = engineer_name[0]
ASCII = ord(first_letter)

verification_hash = (ASCII * SEED_NUM) - loops

exec_log = f"'{first_letter}' -> {ASCII}, {ASCII}*{SEED_NUM}-{loops}={verification_hash}"

print(f"First Letter ASCII (A): {ASCII}")
print(f"SEED_NUM (B): {SEED_NUM}")
print(f"Total Loops (C): {loops}")
print(f"Verification Hash: {verification_hash}")
print(f"Execution Log: {exec_log}")
```

```python
# [EXERCISE 3]
correct_password = "bakachesterto"
attempts = 0
log_tracker = []

while True:
    user_input = input("Enter password: ")
    attempts += 1

    if user_input == correct_password:
        log_tracker.append(f"Try {attempts}: Match")
        print("Access Granted")
        break
    else:
        log_tracker.append(f"Try {attempts}: Fail")
        print("Incorrect Password. Try Again.")

exec_log = ", ".join(log_tracker) 

print(f"Input Attempts: {attempts}")
print(f"Attempts Before Success: {attempts}")
print(f"Execution Log:{exec_log}")
```
